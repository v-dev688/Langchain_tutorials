# RAG CHATBOT

```python
from langchain_community.document_loaders import PyPDFLoader

loader=loader = PyPDFLoader(
    r"C:\Users\vdrss\Downloads\Langchain_tutorials\3_SIMPLE_RAG_APP\CSA dcouments\2014-CSA-Code-of-Ethics.pdf"
)
documents=loader.load()

context ="\n\n".join(doc.page_content for doc in documents)
```

```python
from langchain_core.prompts import PromptTemplate

prompt=PromptTemplate(
    input_variables=["context","question"],
    template="""
    You are an intelligent assistant trained to understand and provide information based on the context provided
    Context: {context}
    Question: {question}

    Your responses should directly address the question using the information from the context,
    while being clear and context
"""

)
```

```python
formatted_prompt= prompt.format(
    context=context,
    question="Under the CSA Code of Ethics, what three steps must a CSA stakeholder take when a direct or indirect conflict of interest arises?"
)
result=llm.invoke(formatted_prompt)
result.content
```

```python
for m in llm.stream(formatted_prompt):
    print(m.content,end="",flush=True)
```
